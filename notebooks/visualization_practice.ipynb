```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
 
spotify_df = pd.read_csv("../data/Cleaned_Spotify_2024_Global_Streaming_Data.csv")
spotify_df.head()
 
```

```python
# group by genre and calculate average streams
avg_streams = spotify_df.groupby("Genre")["Total Streams (Millions)"].mean().sort_values(ascending=False)
print(avg_streams)
```

```python
plt.bar(avg_streams.index, avg_streams.values)
```

```python
plt.bar(avg_streams.index, avg_streams.values)
plt.xlabel("Genre")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Average Total Streams by Genre")

plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.bar(avg_streams.index, avg_streams.values)
plt.xlabel("Genre")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Average Total Streams by Genre")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.bar(avg_streams.index, avg_streams.values)


ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.xlabel("Genre")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Average Total Streams by Genre")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
avg_artists = spotify_df.groupby("Artist")["Total Streams (Millions)"].mean().sort_values(ascending=False)
avg_artists
```

```python
plt.figure(figsize=(10, 6))
colors = ["#8D9FE9"]*len(avg_artists)

max_index = avg_artists.idxmax()
highlight_position = avg_artists.index.get_loc(max_index)
colors[highlight_position] = "#97282E"

plt.bar(avg_artists.index, avg_artists.values, color="#8D9FE9", width=0.8)
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.xlabel("Artist")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Average Total Streams by Artist")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

special_artist = "Ariana Grande"
colors = ["#f030ad" if artist == special_artist else "gray" for artist in avg_artists.index]

plt.bar(avg_artists.index, avg_artists.values, color="#8D9FE9", width=0.8)
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.xlabel("Artist")
plt.ylabel("Average Total Streams (Millions)")
plt.title("Ariana Grande Ranks 2nd in Average Total Streams")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
yearly_average = spotify_df.groupby("Release Year")["Total Streams (Millions)"].mean().round(2)
yearly_average
```

```python
plt.figure(figsize=(10, 6))
plt.plot(
    yearly_average.index,         
    yearly_average.values,       
    marker="+",
    linestyle="--",
    color="#20a32b",
    linewidth=0.5
)

plt.xlabel("Release Year")
plt.ylabel("Average Total Streams")
plt.title("Did Listerners Stream Less in 2022?", fontsize=14, pad=20)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.axvline(
        2022, 
        color="#063b0a",
        linewidth= 0.5, 
        alpha=0.7,
)
ax.text(
    2022+0.05,
    2225, 
    "Did in Total Streams",
    color="#063b0a",
    fontsize=10,
)

plt.ylim(0, 3500)

plt.show()
```

```python
category = "Platform Type"
values = "Total Streams (Millions)"

platform_totals = spotify_df.groupby(category)[values].sum()
platform_totals
```

```python
plt.figure(figsize=(8,8))
plt.pie(
    platform_totals,
    labels=platform_totals.index,
    startangle=110,
    colors=["#1db954", "#d1d5d1"],
    autopct="%1.1f%%",
)

plt.title("Hear Even Split Between Premium and Free Users")

plt.legend(
    labels=platform_totals.index,
    loc="lower left",
    title="Platform",
)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,8))

wedges, texts, autotexts = ax.pie(
    platform_totals,
    labels=platform_totals.index,
    startangle=110,
    colors=["#d1d5d1", "#1db954"],
    autopct="%1.1f%%",
)

ax.set_title(
    "Near Even Split Between Premium and Free Users",
    fontsize=30,
    pad=20,
    loc="center"
    )
for autotexts in autotexts:
      autotexts.set_fontsize(30)
      autotexts.set_color("white")
      autotexts.set_weight("bold")

ax.legend(
    wedges,
    platform_totals.index,
    loc="lower center",
    bbox_to_anchor=(0.5, 0.92),
    ncol=2
)
plt.show()
```

```python
numeric_spotify_df = spotify_df.select_dtypes(include=[np.number])
corr = numeric_spotify_df.corr()

im = plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(im, label="Pearson Correlation")

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha="right")
plt.yticks(range(len(corr.columns)), corr.columns)

plt.show()
```
